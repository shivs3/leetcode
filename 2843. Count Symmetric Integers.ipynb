# 2843. Count Symmetric Integers

```python
low = 1
high = 100
```

```python
import math
count=0
for num in range(low, high):
    length = len(str(num))
    n = math.log2(length)
    if n%1 != 0 or n<1:
        continue
    print(num)
    print(n)
    sum1,sum2 = 0,0
    for i in range(int(n)):
        sum1+=num%10
        num=num//10
    print(sum1)
    for i in range(int(n)):
        sum2 += num%10
        num = num//10
    print(sum2)
    if sum1 == sum2:
        count+=1
```
