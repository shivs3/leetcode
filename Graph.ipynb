### Implementing graphs from scratch using classes

```python
class Graph:
    def __init__(self):
        self.adj_list = {}

    #define a function to ad node
    def addNode(self, node):
        if node not in self.adj_list:
            self.adj_list[node] = []

    # A function to ad Edge for Undirected Graph
    def addEdge(self, node1, node2):
        if node1 not in self.adj_list:
            self.addNode(node1)
        if node2 not in self.adj_list:
            self.addNode(node2)
        self.adj_list[node1].append(node2)
        self.adj_list[node2].append(node1)

    # A Function to display
    def display(self):
        for node in self.adj_list:
            print(f'{node} : {self.adj_list[node]}')
```

```python
class Graph:
    def __init__(self):
        self.adj_list = {}

    #define a function to ad node
    def addNode(self, node):
        if node not in self.adj_list:
            self.adj_list[node] = []

    # A function to ad Edge for Undirected Graph
    def addEdge(self, node1, node2):
        if node1 not in self.adj_list:
            self.addNode(node1)
        if node2 not in self.adj_list:
            self.addNode(node2)
        self.adj_list[node1].append(node2)
        self.adj_list[node2].append(node1)

    # A Function to display
    def display(self):
        for node in self.adj_list:
            print(f'{node} : {self.adj_list[node]}')
    
    # Breadth First Search
    def bfs(self, start):
        q = [start]
        visited = set()
        while q:
            node = q.pop(0)
            if node not in visited:
                print(node, end = ' ')
                visited.add(node)
                q.extend(self.adj_list[node])
            
```

```python
g = Graph()
g.addEdge('A','C')
g.addEdge('A','B')
g.addEdge('B','D')
g.addEdge('D','C')
g.display()
```

```python
g.bfs('D')
```

## DFS

```python
class Graph:
    def __init__(self):
        self.adj_list = {}

    #define a function to ad node
    def addNode(self, node):
        if node not in self.adj_list:
            self.adj_list[node] = []

    # A function to ad Edge for Undirected Graph
    def addEdge(self, node1, node2):
        if node1 not in self.adj_list:
            self.addNode(node1)
        if node2 not in self.adj_list:
            self.addNode(node2)
        self.adj_list[node1].append(node2)
        self.adj_list[node2].append(node1)

    # A Function to display
    def display(self):
        for node in self.adj_list:
            print(f'{node} : {self.adj_list[node]}')
    
    # Depth First Search
    def dfs(self, start, visited = None):
        if visited is None:
            visited = set()
        print(start, end = ' ')
        visited.add(start)
        for nei in self.adj_list[start]:
            if nei not in visited:
                self.dfs(nei, visited)
        
```

```python
g = Graph()
g.addEdge('A','C')
g.addEdge('A','B')
g.addEdge('B','D')
g.addEdge('D','C')
g.display()
```

```python
g.dfs('B')
```

```python
class Graph:
    def __init__(self):
        self.adj_list = {}

    def addNode(self, node):
        if node not in self.adj_list:
            self.adj_list[node] = []
    
    def addEdge(self, node1, node2):
        if node1 not in self.adj_list:
            self.addNode(node1)
        if node2 not in self.adj_list:
            self.addNode(node2)
        self.adj_list[node1].append(node2) #Directed Graph

    # A Function to display
    def display(self):
        for node in self.adj_list:
            print(f'{node} : {self.adj_list[node]}')

    def dfs(self, node, visited = None):
        if visited is None:
            visited = set()
        visited.add(node)
        for neighbors in self.adj_list[node]:
            if neighbors not in visited:
                self.dfs(neighbors, visited)
            else:
                return False
        return True
```

```python
prerequisites = [[1,0]]
g = Graph()
def test(prerequisites,numCourses):
    for x, y in prerequisites:
        g.addEdge(x,y)
    for i in range(numCourses):
        if g.dfs(i):
            return False
    return True
```
