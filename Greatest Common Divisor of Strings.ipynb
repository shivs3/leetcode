# Greatest Common Divisor of Strings

```python
'AB'*3
```

```python
str1 = "ABCABC"
str2 = "ABC"
res = ''
i=1
while i<=len(str1) and i<=len(str2):
    if str1[:i]*(len(str1)//i) == str1 and str1[:i]*(len(str2)//i) == str2:
        res+=str1[:i]
    i+=1
```
