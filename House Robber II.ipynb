# House Robber II

```python
def helper(nums):
    rob1,rob2=0,0
    for num in nums:
        temp=max(num+rob1,rob2)
        rob1=rob2
        rob2=temp
    return rob2
```

```python
def money(nums):
    return max(nums[0], helper(nums[:-1]),helper(nums[1:]))
```

```python
money([2,3,2])
```

```python
money([1,2,3,1])
```
